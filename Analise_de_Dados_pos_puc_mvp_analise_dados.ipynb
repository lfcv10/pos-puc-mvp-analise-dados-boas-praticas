# 1. Problema e perguntas

## 1.1 Contexto e objetivo

Os acidentes de trânsito nas rodovias federais brasileiras representam um problema de segurança e saúde pública, com impactos sociais e econômicos.

O projeto utilizará dados da Polícia Rodoviária Federal (PRF) de acidentes ocorridos em rodovias no ano de 2025 para analisar a distribuição temporal e geográfica dos acidentes, investigando características como horário, dia da semana, tipo de acidente, condições meteorológicas e quantidade de vítimas.

O objetivo é identificar padrões e associações entre as características dos acidentes e sua gravidade, por meio de uma análise exploratória e descritiva, sem a perspectiva de realizar uso preditivo futuro.

## 1.2 Perguntas ou hipóteses

- **H1:** Como a quantidade e a gravidade dos acidentes variam conforme o horário, dia da semana e época do ano?
- **H2:** Quais características dos acidentes estão mais associadas à ocorrência de vítimas fatais?
- ** H3:** Quais são os estados e rodovias mais perigosos? Quais trechos?


OBS: Foi selecionado o recorte do ano de 2025 por se tratar do último ano com dados fechados, permitindo uma visão próxima
